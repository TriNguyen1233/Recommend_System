```python
import os
import pandas as pd
import joblib
import torch
import torch.nn as nn
from sklearn.preprocessing import StandardScaler
import ast
# 1. Đọc dữ liệu
Electronics_Rating = pd.read_csv("./train_data(17-18)/Electronics.csv")
Electronics_Product = pd.read_csv("./train_data(17-18)/amazon_product_data.csv")


Electronics_Product["price"]=Electronics_Product['price'].astype(str).str.replace(r'[^\d.]', '', regex=True)
Electronics_Product["price"]=pd.to_numeric(Electronics_Product["price"], errors='coerce')
median_val = Electronics_Product["price"].median()
Electronics_Product["price"]=Electronics_Product["price"].fillna(median_val)
price_embedding=10
# Product_Rating_Data['price_bucket'] = pd.qcut(Product_Rating_Data['price'], q=price_embedding, labels=False)
# Product_Rating_Data['price_bucket'] = Product_Rating_Data['price_bucket'].fillna(0).astype('int32')
# log transform
Electronics_Product["price_log"] = np.log1p(Electronics_Product["price"])


# scale

user_avg = Electronics_Rating.groupby("user_id")["rating"].mean()
user_avg = (user_avg - 3) / 2
Electronics_Rating['user_avg_rating'] = Electronics_Rating['user_id'].map(user_avg)
user_var = Electronics_Rating.groupby("user_id")["rating"].var()
Electronics_Rating['user_rating_var'] = Electronics_Rating['user_id'].map(user_var)

Electronics_Rating['verified_purchase'] = (
    Electronics_Rating['verified_purchase']
    .astype(str)
    .str.lower()
    .map({'true': 1, 'false': 0})
)
Electronics_Rating['verified_purchase'] = Electronics_Rating['verified_purchase'].fillna(0)


# tinh trung binh rating
Electronics_Rating['rating'] = pd.to_numeric(Electronics_Rating['rating'], errors='coerce')



```

```python
import ast
import pandas as pd

# 1. Định nghĩa lại hàm (đặt tên khác để tránh trùng với biến)
def get_details(details_str):
    # Chuyển string thành dict an toàn
    try:
        details = ast.literal_eval(details_str) if isinstance(details_str, str) else {}
    except:
        details = {}
        
    if not isinstance(details, dict):
        return pd.Series({'brand': 'unk', 'color': 'unk', 'weight': 'unk', 'dimensions': 'unk','Country of Origin': 'unk'})
    
    return pd.Series({
        'brand': details.get('Brand', 'unk'),
        'color': details.get('Color', 'unk'),
        'weight': details.get('Item Weight', 'unk'),
        'dimensions': details.get('Package Dimensions', 'unk'),
        'Country of Origin': details.get('Country of Origin', 'unk')
    })


# 2. Trích xuất vào một biến tạm
new_features = Electronics_Product['details'].apply(get_details)

# 3. Xóa các cột cũ nếu đã tồn tại để tránh trùng tên khi concat
cols_to_drop = ['brand', 'color', 'weight', 'dimensions', 'Country of Origin']
Electronics_Product = Electronics_Product.drop(columns=[c for c in cols_to_drop if c in Electronics_Product.columns])

# 4. Nối dữ liệu mới
Electronics_Product = pd.concat([Electronics_Product, new_features], axis=1)

brand_counts = Electronics_Product['brand'].value_counts()
tail_brands = brand_counts[brand_counts <= 2].index
Electronics_Product["brand"]=Electronics_Product["brand"].apply(lambda x: 'unk' if x in tail_brands else x)

category_counts = Electronics_Product['last_category'].value_counts()
tail_categories = category_counts[category_counts <= 2].index
Electronics_Product["last_category"]=Electronics_Product["last_category"].apply(lambda x: 'unk' if x in tail_categories else x)

color_counts = Electronics_Product['color'].value_counts()
tail_colors = color_counts[color_counts <= 2].index
Electronics_Product["color"]=Electronics_Product["color"].apply(lambda x: 'unk' if x in tail_colors else x)

country_counts = Electronics_Product['Country of Origin'].value_counts()
tail_countries = country_counts[country_counts <= 2].index
Electronics_Product["Country of Origin"]=Electronics_Product["Country of Origin"].apply(lambda x: 'unk' if x in tail_countries else x)

store_counts= Electronics_Product['store'].value_counts()
tail_stores = store_counts[store_counts <= 2].index
Electronics_Product["store"]=Electronics_Product["store"].apply(lambda x: 'unk' if x in tail_stores else x)

# 5. Ép kiểu về string để đảm bảo LabelEncoder không gặp lỗi dữ liệu hỗn hợp
# Và sử dụng .values để đưa về mảng 1 chiều chuẩn
```

```python
scaler = StandardScaler()
Electronics_Product["price_scaled"] = scaler.fit_transform(Electronics_Product[["price_log"]])
Electronics_Rating["user_code"] = user_encoder.fit_transform(Electronics_Rating["user_id"])

Electronics_Product["category_code"] = final_category_encoder.fit_transform(Electronics_Product["last_category"])
Electronics_Product['main_category'] = main_category_encoder.fit_transform(Electronics_Product['main_category'])

Electronics_Product['parent_asin_code']=parent_encoder.fit_transform(Electronics_Product['parent_asin'])

Electronics_Rating["asin_code"] = item_encoder.fit_transform(Electronics_Rating["asin"])

Electronics_Product['country_code'] = country_encoder.fit_transform(Electronics_Product['Country of Origin'])
Electronics_Product['store_code']=store_encoder.fit_transform(Electronics_Product['store'])
Electronics_Product["brand_code"] = brand_encoder.fit_transform(Electronics_Product["brand"].astype(str))
Electronics_Product["color_code"] = color_encoder.fit_transform(Electronics_Product["color"].astype(str))
```

```python
user_history = Electronics_Rating.groupby('user_code')['asin_code'].apply(list).to_dict()

def get_recent_history(row, max_len=5):
    uid = row['user_code']
    current_item = row['asin_code']
    
    full_hist = user_history.get(uid, [])
    
    try:
        idx = full_hist.index(current_item)
        sub_hist = full_hist[:idx]
    except ValueError:
        sub_hist = []
        
    if len(sub_hist) > max_len:
        sub_hist = sub_hist[-max_len:]
        
    padded_hist = [0] * (max_len - len(sub_hist)) + sub_hist
    return padded_hist

Electronics_Rating['history_list'] = Electronics_Rating.apply(get_recent_history, axis=1)
```

```python
import numpy as np
import torch

edge_index_np = np.array([
    Product_Rating_Data['user_code'].values,
    Product_Rating_Data['asin_code'].values + num_users
])

edge_index = torch.from_numpy(edge_index_np).long()

edge_index = torch.cat([
    edge_index,
    edge_index.flip(0)
], dim=1)

edge_index = edge_index.to(device)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import LGConv

class Neural_Network(nn.Module):
    def __init__(self, num_users, num_items, num_brand, num_category, num_main_category,
                 num_color, num_store, num_parent_asin, num_country, edge_index):
        super(Neural_Network, self).__init__()
        
        # ===== 1. Embeddings Gốc =====
        self.user_embedding = nn.Embedding(num_users, 64) 
        self.item_embedding = nn.Embedding(num_items, 64)

        self.convs = nn.ModuleList([LGConv() for _ in range(2)]) # Đã sửa thành 2 lớp theo tối ưu trước

        # ===== 2. Embeddings Thuộc Tính Phân Loại =====
        self.brand_embedding = nn.Embedding(num_brand, 4)           
        self.category_emb = nn.Embedding(num_category, 4)           
        self.main_category_emb = nn.Embedding(num_main_category, 2) 
        self.color_embedding = nn.Embedding(num_color, 2)           
        self.store_embedding = nn.Embedding(num_store, 4)           
        self.parent_asin_embedding = nn.Embedding(num_parent_asin, 8) 
        self.country_embedding = nn.Embedding(num_country, 2)       
        
        # Bias cho Matrix Factorization bổ trợ
        self.user_bias = nn.Embedding(num_users, 1)
        self.item_bias = nn.Embedding(num_items, 1)
        self.global_bias = nn.Parameter(torch.zeros(1))

    
        self.gru = nn.GRU(input_size=64, hidden_size=64, batch_first=True, num_layers=1)

        # ===== 3. Các Lớp Tuyến Tính Cho Dữ Liệu Số =====
        self.price_fc = nn.Linear(1, 8)
        self.avg_rating_fc = nn.Linear(1, 8)
        self.rating_num_fc = nn.Linear(1, 8)
        self.user_stat_fc = nn.Linear(2, 8)

        # ===== 4. Dropouts Chống Overfitting =====
        self.id_dropout = nn.Dropout(p=0.2)
        self.feat_dropout = nn.Dropout(p=0.2)
        self.num_dropout = nn.Dropout(p=0.2)
        self.inter_dropout = nn.Dropout(p=0.2)
        
        self.register_buffer("edge_index", edge_index)
        
        # Các cổng Gating kết hợp GCN (Đã sửa kích thước đầu ra thành 64 theo tối ưu trước)
        self.gate_u = nn.Linear(128, 64)
        self.gate_i = nn.Linear(128, 64)
     
        input_dim = 252

        self.mlp = nn.Sequential(
            nn.Linear(input_dim, 128), # Đã mở lại tầng 128 theo mô hình kim tự tháp
            nn.BatchNorm1d(128),
            nn.LeakyReLU(0.1),
            nn.Dropout(0.3),

            nn.Linear(128, 64),
            nn.BatchNorm1d(64),
            nn.LeakyReLU(0.1),
            nn.Dropout(0.2),

            nn.Linear(64, 32),
            nn.LeakyReLU(0.1)
        )

        self.output_layer = nn.Linear(32, 1)

    def forward(self, user_id, item_id, history_item_ids, category_code, brand_code, price_value, 
                avg_rating, rating_number, main_category, user_avg, user_var, color_code, 
                store_code, parent_asin_code, country_code):
        """
        history_item_ids: LongTensor dạng [batch_size, seq_len] chứa các ID sản phẩm quá khứ của User
        """
        
        # Lan truyền đồ thị LightGCN (Đã bỏ chuẩn hóa inline ở từng tầng theo tối ưu trước)
        x_all = torch.cat([
            self.user_embedding.weight,
            self.item_embedding.weight
        ], dim=0)

        embs = [x_all]
        for conv in self.convs:
            x_all = conv(x_all, self.edge_index)
            x_all = torch.clamp(x_all, -10.0, 10.0) 
            embs.append(x_all)

        final_emb = torch.stack(embs, dim=0).mean(dim=0)
        final_emb = F.normalize(final_emb, p=2, dim=-1, eps=1e-8) # Chuẩn hóa 1 lần duy nhất sau khi lấy trung bình

        num_users = self.user_embedding.num_embeddings
        all_user_gcn_emb = final_emb[:num_users]
        all_item_gcn_emb = final_emb[num_users:]

        # Lấy Embedding tĩnh cho Target User và Target Item
        u_emb = F.normalize(self.user_embedding(user_id), p=2, dim=-1, eps=1e-8)
        i_emb = F.normalize(self.item_embedding(item_id), p=2, dim=-1, eps=1e-8)

        u_gcn = F.normalize(all_user_gcn_emb[user_id], p=2, dim=-1, eps=1e-8)
        i_gcn = F.normalize(all_item_gcn_emb[item_id], p=2, dim=-1, eps=1e-8)

        # Gating linh hoạt theo từng phần tử đặc trưng (Element-wise Gating)
        g_u = torch.sigmoid(self.gate_u(torch.cat([u_emb, u_gcn], dim=-1)))
        g_i = torch.sigmoid(self.gate_i(torch.cat([i_emb, i_gcn], dim=-1)))

        u = self.id_dropout(u_emb + g_u * u_gcn)
        i = self.id_dropout(i_emb + g_i * i_gcn)

        # ===== XỬ LÝ CHUỖI HÀNH VI BẰNG TẦNG GRU =====
        # Lấy các vector embedding chất lượng cao của chuỗi sản phẩm quá khứ từ ma trận Item GCN
        # history_item_ids shape: [batch_size, seq_len] -> history_embs shape: [batch_size, seq_len, 64]
        history_embs = all_item_gcn_emb[history_item_ids]
        
        # Đưa qua mạng GRU hành vi
        # hn chứa hidden state tại bước thời gian cuối cùng: shape [1, batch_size, 64]
        _, hn = self.gru(history_embs)
        u_seq = hn.squeeze(0) # Chuyển về dạng phẳng [batch_size, 64]
        u_seq = self.id_dropout(u_seq) # Dropout chống học vẹt chuỗi hành vi

        # Các thuộc tính phân loại khác
        b = self.feat_dropout(self.brand_embedding(brand_code))
        c = self.feat_dropout(self.category_emb(category_code))
        m_cat = self.feat_dropout(self.main_category_emb(main_category))
        color_emb = self.feat_dropout(self.color_embedding(color_code)) 
        store_emb = self.feat_dropout(self.store_embedding(store_code)) 
        parent_asin_emb = self.feat_dropout(self.parent_asin_embedding(parent_asin_code))  
        country_emb = self.feat_dropout(self.country_embedding(country_code))  

        # Chuẩn hóa dữ liệu số thực
        price_safe = torch.log1p(torch.clamp(price_value, min=0.0)).clamp(0.0, 10.0)
        rating_safe = torch.clamp(avg_rating, min=0.0, max=5.0)
        rating_num_safe = torch.log1p(torch.clamp(rating_number, min=0.0))
        
        user_avg_safe = torch.clamp(user_avg, min=0.0, max=5.0)
        user_var_safe = torch.log1p(torch.clamp(user_var, min=0.0))

        p = self.num_dropout(torch.tanh(self.price_fc(price_safe.unsqueeze(-1))))
        ua = self.num_dropout(torch.tanh(self.avg_rating_fc(rating_safe.unsqueeze(-1))))
        ia = self.num_dropout(torch.tanh(self.rating_num_fc(rating_num_safe.unsqueeze(-1))))
        u_stats = self.num_dropout(self.user_stat_fc(torch.stack([user_avg_safe, user_var_safe], dim=-1)))

        # Tính toán ma trận tương tác tĩnh (Cross-Features)
        dot = torch.sum(u * i, dim=-1, keepdim=True)
        cos = F.cosine_similarity(u, i, dim=-1, eps=1e-8).unsqueeze(-1)
        
  
        x = torch.cat([
            u, i, u_seq,
            dot, cos,
            p, ua, ia, u_stats,
            country_emb, color_emb, store_emb, m_cat, parent_asin_emb, b, c
        ], dim=-1)

        # Đưa qua mạng MLP giải mã phi tuyến sâu
        mlp_out = self.mlp(x)
        out = self.output_layer(mlp_out).squeeze(-1)
        
        # ===== KÍCH HOẠT BIAS THAM SỐ CHẾT =====
        # Cộng thêm bias cục bộ của User, Item và Global bias trước khi kích hoạt Sigmoid
        out = out + self.user_bias(user_id).squeeze(-1) + self.item_bias(item_id).squeeze(-1) + self.global_bias
        
        return torch.sigmoid(out) * 4.0 + 1.0
```

```python
import torch
from torch.utils.data import Dataset



class Electronics_RatingDataset(Dataset):
    def __init__(self, df):
        self.user_code=torch.tensor(df["user_code"].values)
        self.asin = torch.tensor(df["asin_code"].values)
        self.ratings = torch.tensor(df["rating"].values, dtype=torch.float32)
        self.category_code=torch.tensor(df["category_code"].values, dtype=torch.long)
        self.brand_code=torch.tensor(df["brand_code"].values, dtype=torch.long)
        self.price_values=torch.tensor(df["price_scaled"].values, dtype=torch.float32)
        self.avg_rating=torch.tensor(df["average_rating"].values, dtype=torch.float32)
        self.rating_number=torch.tensor(df["rating_number"].values, dtype=torch.float32)
        self.user_rating_avg=torch.tensor(df["user_avg_rating"].values, dtype=torch.float32)
        self.user_rate_var = torch.tensor(df['user_rating_var'].values, dtype=torch.float32)
        self.main_category=torch.tensor(df["main_category"].values, dtype=torch.float32)
        self.color_code=torch.tensor(df["color_code"].values, dtype=torch.long)
        self.store_code=torch.tensor(df["store_code"].values, dtype=torch.long)
        self.parent_asin_code=torch.tensor(df["parent_asin_code"].values, dtype=torch.long)
        self.country_code=torch.tensor(df["country_code"].values, dtype=torch.long)
        self.history_items=torch.tensor(df["history_list"].tolist(), dtype=torch.long)
      
    def __len__(self):
        return len(self.ratings)

    def __getitem__(self, idx):
        return (self.user_code[idx],self.asin[idx],self.history_items[idx],self.category_code[idx],self.brand_code[idx], self.ratings[idx],self.price_values[idx]
        ,self.avg_rating[idx],self.rating_number[idx],self.main_category[idx],self.user_rating_avg[idx]
        ,self.user_rate_var[idx],self.color_code[idx],self.store_code[idx],self.parent_asin_code[idx],self.country_code[idx], 
       )
```

```python
import torch


def train(model, train_loader, optimizer, criterion, device):
    model.train()
    total_loss = 0
    total_mae = 0
    total = 0
    correct=0
    diff_threshold = 0.5  # dùng để tính accuracy kiểu "gần đúng"
    # model.update_gcn()

    for (user_batch, item_batch,history_batch, category_batch, brand_batch, rating_batch, price_batch,avg_rating,
    rating_number,main_category,user_avg,user_var,color_batch,store_batch,parent_asin_batch,country_batch) in train_loader:

        user_batch = user_batch.to(device).long()
        item_batch = item_batch.to(device).long()
        history_batch = history_batch.to(device).long()
        category_batch = category_batch.to(device).long()
        brand_batch = brand_batch.to(device).long()
        price_batch = price_batch.to(device).float()
        avg_rating=avg_rating.to(device).float()
        rating_number=rating_number.to(device).float()
        color_batch=color_batch.to(device).long()
        store_batch=store_batch.to(device).long()
        parent_asin_batch=parent_asin_batch.to(device).long()
        country_batch=country_batch.to(device).long()
        main_category=main_category.to(device).long()
        user_avg=user_avg.to(device).float()
        user_var=user_var.to(device).float()
        rating_batch = rating_batch.to(device).float()
        

        optimizer.zero_grad()
        prediction = model(user_batch,item_batch,history_batch,category_batch, brand_batch, price_batch,avg_rating,
                           rating_number,main_category,user_avg,user_var,color_batch
                           ,store_batch,parent_asin_batch,country_batch)
        prediction = prediction.squeeze(-1)  # (B, 1) → (B,)

       
        loss = criterion(prediction, rating_batch)

        loss.backward()
        optimizer.step()

        total_loss += loss.item() * rating_batch.size(0)

        # ===== MAE =====
        abs_diff = torch.abs(prediction - rating_batch)
        total_mae += abs_diff.sum().item()

        # ===== "Accuracy" kiểu recommender =====
        correct += (abs_diff <= diff_threshold).sum().item()

        total += rating_batch.size(0)

    avg_loss = total_loss / total
    avg_mae = total_mae / total
    accuracy = (correct / total) * 100  

    return avg_loss, accuracy, avg_mae
```

```python
import torch
import numpy as np

def evaluate(model, test_loader, criterion, device):
    model.eval()
    total_loss = 0
    total_mae = 0
    sq_error_sum = 0
    total = 0
    correct = 0
    diff_threshold = 0.5  # accuracy kiểu "gần đúng"

    with torch.no_grad():
        for (user_batch, item_batch,history_batch, category_batch, brand_batch, rating_batch, price_batch,avg_rating,
        rating_number,main_category,user_avg,user_var,color_batch,store_batch
        ,parent_asin_batch,country_batch) in test_loader:
            user_batch = user_batch.to(device).long()
            item_batch = item_batch.to(device).long()
            history_batch = history_batch.to(device).long()
            category_batch = category_batch.to(device).long()
            brand_batch = brand_batch.to(device).long()
            price_batch = price_batch.to(device).float()
            true_rating = rating_batch.to(device).float()
            avg_rating=avg_rating.to(device).float()
            rating_number=rating_number.to(device).float()
            user_avg=user_avg.to(device).float()
            user_var=user_var.to(device).float()
            main_category=main_category.to(device).long()
            color_batch=color_batch.to(device).long()
            store_batch=store_batch.to(device).long()
            parent_asin_batch=parent_asin_batch.to(device).long()
            country_batch=country_batch.to(device).long()
            predictions = model(user_batch, item_batch, history_batch, category_batch, brand_batch, price_batch,avg_rating,rating_number,
                                main_category,user_avg,user_var,color_batch
                                ,store_batch,parent_asin_batch,country_batch)
            predictions = predictions.squeeze(-1)


            loss = criterion(predictions, true_rating)
            total_loss += loss.item() * true_rating.size(0)

            # ===== MAE =====
            abs_diff = torch.abs(predictions - true_rating)
            total_mae += abs_diff.sum().item()

            # ===== RMSE =====
            sq_error_sum += (abs_diff ** 2).sum().item()

            # ===== Accuracy kiểu recommender =====
            correct += (abs_diff <= diff_threshold).sum().item()

            total += true_rating.size(0)

    avg_loss = total_loss / total
    avg_mae = total_mae / total
    rmse = np.sqrt(sq_error_sum / total)
    accuracy = (correct / total) * 100  # metric phụ

    return rmse, accuracy, avg_mae
```

```python
import os
import torch

num_epochs = 500

# ===== 1. THAY ĐỔI TIÊU CHÍ ĐÁNH GIÁ SANG RMSE =====
best_rmse = float('inf')  # Chuyển từ best_accuracy sang best_rmse
best_accuracy = 0  
best_mae = float('inf')

early_stop_patience = 15
no_improve_epochs = 0

print("-" * 65)
print(f"epoch: | train_loss:| train_acc| test_rmse | test_acc | diff: ")

for epoch in range(num_epochs):
    # Thực hiện Train và Đánh giá
    train_loss, train_acc, train_mae = train(model, train_loader, optimizer, criterion, device)
    test_rmse, test_acc, test_mae = evaluate(model, test_loader, criterion, device)
    
    # ===== 2. ĐỒNG BỘ SCHEDULER =====
    # Nếu bạn dùng ReduceLROnPlateau, dòng này chuẩn 100%
    scheduler.step(test_mae)
    
    # ===== 3. SỬA LOGIC LƯU CHECKPOINT THEO RMSE THẤP NHẤT =====
    if test_mae < best_mae:
        best_rmse = test_rmse
        best_accuracy = test_acc
        best_mae = test_mae
        no_improve_epochs = 0  # Reset bộ đếm khi tìm thấy điểm tối ưu mới

        checkpoint = {
            'epoch': epoch + 1,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'test_acc': test_acc,
            'test_rmse': test_rmse
        }
        
        save_dir = "./content/weights"
        os.makedirs(save_dir, exist_ok=True)
        save_path = os.path.join(save_dir, "best_model.pth")
        torch.save(checkpoint, save_path)
        
        # In thông báo dựa trên vị vua RMSE
        print(f"--> Đã lưu model mới tốt nhất tại Epoch {epoch+1} với mae: {test_mae:.4f}")
        
    else:
        no_improve_epochs += 1
        if no_improve_epochs >= early_stop_patience:
            print("=== Early stopping triggered! ===")
            break
            
    # Giữ nguyên định dạng in log của bạn cho mỗi Epoch
    print(f"{epoch+1:<6} | {train_loss:<10.4f} | {train_acc:<8.2f}% | {test_rmse:<10.4f} | {test_acc:<8.2f}% | diff: {test_mae:.3f} sao")
```
